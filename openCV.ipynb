# Color Recognition Using OpenCV

This project detects and recognizes Red, Green, and Blue colors in real-time using a webcam.

The project uses the HSV color space to identify colors and OpenCV to draw bounding boxes around detected colored objects.

## How Color Recognition Works

1. Open the webcam using OpenCV.
2. Capture video frames in real-time.
3. Convert each frame from BGR to HSV color space.
4. Define HSV ranges for Red, Green, and Blue colors.
5. Create a mask for each color.
6. Find the contours of colored objects.
7. Draw a bounding box around each detected object.
8. Display the detected color name on the screen.

# Press **Q** to close the program.

```python
import cv2
import numpy as np

# Select camera:
# 0 = Default/Front camera
# 1 = Secondary/Rear camera (depending on device)
cap = cv2.VideoCapture(1)

# Check if the webcam opened successfully
if not cap.isOpened():
    print("Error: Could not open webcam.")
    raise SystemExit

while True:
    # Capture a frame from the webcam
    ret, frame = cap.read()

    if not ret:
        print("Error: Could not read frame.")
        break

    # Convert the frame from BGR to HSV
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)

    # ---------------------------
    # Define HSV color ranges
    # ---------------------------

    # Red has two ranges in HSV
    lower_red1 = np.array([0, 120, 70])
    upper_red1 = np.array([10, 255, 255])

    lower_red2 = np.array([170, 120, 70])
    upper_red2 = np.array([180, 255, 255])

    # Green
    lower_green = np.array([35, 50, 50])
    upper_green = np.array([85, 255, 255])

    # Blue
    lower_blue = np.array([100, 150, 50])
    upper_blue = np.array([140, 255, 255])

    # ---------------------------
    # Create masks
    # ---------------------------

    red_mask1 = cv2.inRange(hsv, lower_red1, upper_red1)
    red_mask2 = cv2.inRange(hsv, lower_red2, upper_red2)

    # Combine the two red ranges
    red_mask = cv2.bitwise_or(red_mask1, red_mask2)

    green_mask = cv2.inRange(hsv, lower_green, upper_green)
    blue_mask = cv2.inRange(hsv, lower_blue, upper_blue)

    # Store masks, colors, and labels
    masks = [
        (red_mask, (0, 0, 255), "Red"),
        (green_mask, (0, 255, 0), "Green"),
        (blue_mask, (255, 0, 0), "Blue")
    ]

    # ---------------------------
    # Detect colored objects
    # ---------------------------

    for mask, color, name in masks:

        # Find contours
        contours, _ = cv2.findContours(
            mask,
            cv2.RETR_EXTERNAL,
            cv2.CHAIN_APPROX_SIMPLE
        )

        for contour in contours:

            # Ignore very small objects/noise
            if cv2.contourArea(contour) > 500:

                # Get bounding rectangle
                x, y, w, h = cv2.boundingRect(contour)

                # Draw rectangle around detected color
                cv2.rectangle(
                    frame,
                    (x, y),
                    (x + w, y + h),
                    color,
                    2
                )

                # Display color name
                cv2.putText(
                    frame,
                    name,
                    (x, y - 10),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.7,
                    color,
                    2
                )

    # Display result
    cv2.imshow("Color Recognition", frame)

    # Press Q or ESC to exit
    key = cv2.waitKey(1) & 0xFF

    if key == ord("q") or key == 27:
        break


# Release webcam
cap.release()

# Close all OpenCV windows
cv2.destroyAllWindows()
```

## Camera Selection

This project uses a webcam for real-time color recognition.

The camera index can be changed depending on the available camera devices:

- `0` : Default camera (usually the first/primary camera).
- `1` : Secondary camera (for example, an external webcam or rear camera if available).

In this project, the value is set to `1` to use the rear camera.

If the camera does not open, change the value in:
`cv2.VideoCapture(1)`
to another available camera index.
